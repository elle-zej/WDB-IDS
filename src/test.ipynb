```python
from scapy.all import sniff, IP, TCP, UDP

def handle_packet(packet):
    if IP in packet:
        print("Captured Packet")
        print(f"Source IP: {packet[IP].src}")
        print(f"Destination IP: {packet[IP].dst}")
        print(f"TTL: {packet[IP].ttl}")
        print(f"Protocol Number: {packet[IP].proto}")
        print(f"Packet Length: {packet[IP].len}")

        if TCP in packet:
            print("Layer 4 Protocol: TCP")
            print(f"Source Port: {packet[TCP].sport}")
            print(f"Destination Port: {packet[TCP].dport}")
            print(f"TCP Flags: {packet[TCP].flags}")

        elif UDP in packet:
            print("Layer 4 Protocol: UDP")
            print(f"Source Port: {packet[UDP].sport}")
            print(f"Destination Port: {packet[UDP].dport}")

        print("-" * 50)

sniff(prn=handle_packet, store=False, count=5)
```
